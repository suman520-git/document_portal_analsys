```python
import logging
import os
from datetime import datetime
```

```python
os.getcwd()
```

```python
logs_dir=os.path.join(os.getcwd(),"logs")
logs_dir
```

```python
os.makedirs(logs_dir, exist_ok=True)
```

```python
LOG_FILE = f"{datetime.now().strftime('%m_%d_%Y_%H_%M_%S')}.log"
```

```python
datetime.now()
```

```python
#this is a log file(name of the log file would current data and time)
LOG_FILE
```

```python
#today
# with today's date log file will be created

```

```python
LOG_FILE_PATH=os.path.join(logs_dir, LOG_FILE)
LOG_FILE_PATH
```

```python
logging.basicConfig(
    filename=LOG_FILE_PATH,
    format="[ %(asctime)s ] %(levelname)s %(name)s (line:%(lineno)d) - %(message)s",
    level=logging.INFO,
)
```

```python
logger=logging.getLogger("DocumentPortal")
```

```python
logger.info("This is an info message")
```

```python
# [ 2026-02-20 20:52:38,221 ] INFO DocumentPortal (line:1) - This is an info message
```

```python
import structlog
log=structlog.get_logger()
```

```python
log.error("Failed to process PDF", error="File not found", user_id=123)
```
